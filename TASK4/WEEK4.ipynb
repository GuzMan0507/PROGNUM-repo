```python
import math
import numpy as np
#We define a function, which will contain the game
def game():

    x= input("Input R, P or S (Rock, Paper,Scissors respectively) to play: ")       #With input, we allow the user choose the option he wants

    print(f"You have chosen: {x}")
#answ defines a list that contains the values we want, and indx generate a random value among the ones inside answ
    answ = np.array(['R', 'P', 'S'])
    indx = np.random.randint(0, len(answ), 1)

    print(f"Your oponent has chosen: {answ[indx]}")

#We define the possibilities of the game(win, lose or draw), and we use game() to run the function until the user wins
    if x==answ[indx]:
        print("Draw, play again!")
        game()
    elif  x=='R' and answ[indx]=='P':
        print("Computer wins :(")
        game()
    elif x=='P' and answ[indx]=='S':
        print("Computer wins, try again!")
        game()
    elif x=='S' and answ[indx]=='R':
        print("You lose, try again!")
        game()
    elif x=='R' and answ[indx]=='S':
        print("You win, congratulations!")
    elif x=='P' and answ[indx]=='R':
        print("You win!")
    elif x=='S' and answ[indx]=='P':
        print("You win!!")

game()
```

```python
import math
import numpy as np 
import matplotlib.pyplot as plt

angle_degs = 30 
theta = math.radians(angle_degs)

rotor = math.cos(theta) + 1j*math.sin(theta)
print(rotor, abs(rotor))   

x = np.array([1,3,3,1,1])
y = np.array([1,1,3,3,1])

z= x + 1j * y
z_rot= z*rotor

x_rot = z_rot.real
y_rot = z_rot.imag

plt.figure(figsize=(7,7))
plt.plot(x,y, 'b-o', linewidth=2, label='Original')
plt.plot(x_rot, y_rot,'r-o', linewidth=2, label=f"Rotated {angle_degs}")

plt.axhline(0,color='gray',linewidth=0.5)
plt.axvline(0, color='gray',linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.title(f'Rotation fita={angle_degs}')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.show()
```

```python
import numpy as np

def funcfilter(func,seq):
    x=np.asarray(seq)
    filter= x[x>0]
    return func(filter)

sq = np.array([-100, -10, 0, 10, 100])
result = funcfilter(np.log10, sq)
print("Result log10:", result)

result2=funcfilter(np.log10,100)
print("Result log10:", result2)
```

```python
import numpy as np
from scipy.integrate import quad
from astropy.cosmology import Planck18

H0=Planck18.H0.value
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0

H0si =H0*1000/3.0857e22

def integrand(zp):
    H= H0si*np.sqrt(Om0 *(1+zp)**3+Ode0)
    return(1/((1+zp)*H))
z=2
tL=quad(integrand,0,z)[0]
tLGyr=tL/(3.1557e16)
print(tLGyr)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

np.random.seed(42)
X= np.random.normal(mu_x :=0, sigma_x :=1, 100000)
Y = np.random.normal(mu_y :=2, sigma_y :=1, 100000)
                     
fig, ax = plt.subplots(figsize=(10,6))

ax.hist(X, bins=30, density=True, color='blue', label='X')
ax.hist(Y, bins=30, density=True,alpha=0.3, color='salmon', label='Y')
                     
x_range = np.linspace(-6, 8, 1000)
pdf_X = norm.pdf(x_range, mu_x, sigma_x)
pdf_Y = norm.pdf(x_range, mu_y, sigma_y)  
                     
ax.plot(x_range, pdf_X, 'k:', linewidth=2, label='X(PDF)')
ax.plot(x_range, pdf_Y, 'k--', linewidth=2, label='Y(PDF)')
                     
ax.annotate(r'$\mathcal{N}(0,1)$',
           xy=(0, 0.15), xytext=(-3.5, 0.15),
           color='blue',fontsize=12,
           arrowprops=dict(arrowstyle='->',color='red', lw=2))
                     
ax.annotate(r'$\mathcal{N}(2,1)$',
           xy=(2, 0.15), xytext=(5, 0.15),
           color='red',fontsize=12,
           arrowprops=dict(arrowstyle='->',color='red', lw=2))
                     
ax.set_title('1D histogram + PDF', fontsize=14)
ax.set_xlabel('Values',fontsize=12)
ax.set_ylabel('Probability Density', fontsize=12)
ax.legend(fontsize=11)
ax.set_xlim(-6,8)
ax.set_ylim(0,0.42)
ax.yaxis.grid(True)
ax.set_axisbelow(True)
                     
plt.tight_layout()
plt.show()


fig2, ax2 = plt.subplots(figsize=(7, 6))


h, xedges, yedges = np.histogram2d(X, Y, bins=30)


im = ax2.imshow(h.T, origin='lower', aspect='auto',
                extent=[xedges[0], xedges[-1], yedges[0], yedges[-1]],
                cmap='Blues')

fig2.colorbar(im, ax=ax2)


h_flat = h.flatten()
h_sorted = np.sort(h_flat)[::-1]


cumsum = np.cumsum(h_sorted) / h_sorted.sum()


levels = []
for fraction in [0.68, 0.95, 0.997]:
    idx = np.searchsorted(cumsum, fraction)
    idx = min(idx, len(h_sorted) - 1)
    levels.append(h_sorted[idx])

levels = sorted(levels)  


x_centers = (xedges[:-1] + xedges[1:]) / 2
y_centers = (yedges[:-1] + yedges[1:]) / 2


CS = ax2.contour(x_centers, y_centers, h.T,
                 levels=levels, colors='red', linewidths=1.5)


label_map = {levels[0]: r'$3\sigma$', levels[1]: r'$2\sigma$', levels[2]: r'$1\sigma$'}
fmt = {lev: label_map[lev] for lev in levels}
ax2.clabel(CS, fmt=fmt, fontsize=11, inline=True)


ax2.set_title('2D Histogram + Contour', fontsize=14)
ax2.set_xlabel('X', fontsize=12)
ax2.set_ylabel('Y', fontsize=12)
ax2.set_xlim(-5, 5)
ax2.set_ylim(-2.5, 6.5)
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
