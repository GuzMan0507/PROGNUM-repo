```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def gauss(x, A, x0, sigma, z0):
    return A*np.exp(-(x-x0)**2/(2*sigma**2))+z0
A=float(input("Input the value for A: "))
x0=float(input("Input the value for x0: "))
sigma=float(input("Input the value for sigma: "))
z0=float(input("Input the value for z0: "))
    
low=float(input("Input the lower limit: "))
high=float(input("Input the higher limit: "))

def gauss_int(x):
    return gauss(x, A, x0, sigma, z0)
area, error = quad(gauss_int, low, high)

ini=x0-5*sigma
final= x0+5*sigma

x=np.linspace(ini, final, 500)
y=gauss(x, A, x0, sigma, z0)

plt.figure(figsize=(10,6))
plt.plot(x,y,label='Gaussian')
x_fill = np.linspace(low,high,100)
plt.fill_between(x_fill,gauss(x_fill,A,x0,sigma,z0), color='salmon', alpha=0.5, label=f'Area={area}')
plt.title("Adapatative Graph")
plt.legend()
plt.grid(True)
plt.show()
```
